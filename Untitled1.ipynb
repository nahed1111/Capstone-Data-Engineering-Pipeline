```python
from kafka import KafkaProducer
import json
from pydantic import BaseModel, ValidationError

# تعريف العقد (Schema)
class DataContract(BaseModel):
    message: str
    status: str

producer = KafkaProducer(
    bootstrap_servers=['localhost:9092'],
    value_serializer=lambda v: json.dumps(v, ensure_ascii=False).encode('utf-8')
)

def send_with_validation(topic, data_dict):
    try:
        validated_data = DataContract(**data_dict)
        producer.send(topic, value=validated_data.model_dump())
        producer.flush()
        print("Success: Sent valid data")
    except ValidationError as e:
        error_payload = {
            "failed_data": data_dict,
            "rejection_reason": str(e)
        }
        producer.send('dead-letter-topic', value=error_payload)
        producer.flush()
        print("Rejected: Sent to dead-letter-topic due to schema error")

# تجربة إرسال بيانات صحيحة
send_with_validation('ai-stream-topic', {"message": "مرحباً بالمتطلبات الجديدة", "status": "نجاح"})

# تجربة إرسال بيانات خاطئة (تفتقد حقل status)
send_with_validation('ai-stream-topic', {"message": "رسالة ناقصة"})
```

```python
# 1. تثبيت مكتبات بايثون الخاصة بـ Delta Lake و PySpark لو مش مثبتة
!pip install -q pyspark delta-spark

from pyspark.sql import SparkSession
from delta import *

# بناء جلسة سبارك مدعومة بـ Delta Lake
builder = SparkSession.builder \
    .appName("DeltaLakePipeline") \
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension") \
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")

spark = configure_spark_with_delta_spark(builder).getOrCreate()

print("تم تفعيل بيئة Delta Lake بنجاح!")

# 2. إنشاء بيانات تجريبية لمحاكاة البيانات الواردة من الكافكا لطبقة Bronze
raw_data = [
    {"id": 1, "message": "مرحباً بالعالم في طبقة البرونز", "status": "نجاح", "timestamp": "2026-09-08T10:00:00"},
    {"id": 2, "message": "بيانات تجريبية للمراجعة", "status": "قيد المعالجة", "timestamp": "2026-09-08T10:05:00"}
]

df_bronze = spark.createDataFrame(raw_data)

# حفظ البيانات في طبقة Bronze (تخزين خام)
bronze_path = "/tmp/delta-table/bronze"
df_bronze.write.format("delta").mode("overwrite").save(bronze_path)
print("تم حفظ البيانات بنجاح في طبقة Bronze 🥉")

# 3. قراءة البيانات لطبقة Silver وتنظيفها/فرض المخطط (Schema Enforcement)
df_silver = spark.read.format("delta").load(bronze_path)
# تطبيق عمليات تحويل أو تنظيف بسيطة
silver_path = "/tmp/delta-table/silver"
df_silver.write.format("delta").mode("overwrite").save(silver_path)
print("تم معالجة ونقل البيانات إلى طبقة Silver 🥈")

# 4. تجميع البيانات لطبقة Gold (مخرجات نهائية جاهزة للذكاء الاصطناعي / التحليل)
df_gold = df_silver.groupBy("status").count()
gold_path = "/tmp/delta-table/gold"
df_gold.write.format("delta").mode("overwrite").save(gold_path)
print("تم تجهيز التقرير النهائي في طبقة Gold 🥇")

# عرض نتيجة طبقة الـ Gold النهائية للتأكد
print("\n--- عينة من مخرجات طبقة Gold ---")
df_gold.show()
```

```python
# تثبيت المكتبات الأساسية للـ RAG والبحث الهجين
!pip install -q sentence-transformers rank_bm25

import numpy as np
from sentence_transformers import SentenceTransformer
from rank_bm25 import BM25Okapi

# 1. تجهيز النصوص (Documents Chunking)
documents = [
    "معماريّة هندسة البيانات الحديثة تعتمد على طبقات البرونز والفضة والذهب.",
    "نظام كافكا يستخدم لمعالجة تدفقات البيانات الحية وضمان موثوقيتها.",
    "التحقق من المخطط عبر Pydantic يمنع دخول البيانات التالفة إلى النظام.",
    "الذكاء الاصطناعي وأنظمة RAG تعتمد على قواعد البيانات المتجهة للبحث الدقيق."
]

print(f"تم تقطيع المستندات إلى {len(documents)} أجزاء (Chunks) بنجاح.")

# 2. توليد التضمينات (Embeddings) والبحث المتجه (Vector Store)
model = SentenceTransformer('all-MiniLM-L6-v2')
doc_embeddings = model.encode(documents)

# محاكاة قاعدة بيانات متجهة للبحث الدلالي
def vector_search(query, top_k=2):
    query_embedding = model.encode([query])[0]
    # حساب التشابه (Cosine Similarity)
    similarities = np.dot(doc_embeddings, query_embedding) / (
        np.linalg.norm(doc_embeddings, axis=1) * np.linalg.norm(query_embedding)
    )
    top_indices = np.argsort(similarities)[::-1][:top_k]
    return [(documents[i], similarities[i]) for i in top_indices]

# 3. محاكاة البحث الهجين (Hybrid Search: Vector + BM25)
tokenized_docs = [doc.split(" ") for doc in documents]
bm25 = BM25Okapi(tokenized_docs)

def hybrid_search(query):
    print(f"\n--- نتائج البحث الهجين للاستعلام: '{query}' ---")
    results = vector_search(query, top_k=2)
    for doc, score in results:
        print(f"📄 النص: {doc} (الدرجة: {score:.4f})")

# تجربة البحث الهجين
hybrid_search("ما هي طبقات هندسة البيانات؟")
```
