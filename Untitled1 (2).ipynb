```python
# Stage 1 Continued: Consumer Simulation Check
print("📥 Simulating Consumer Stream Read...")

# Display received items from the local stream simulation
for idx, row in df_stream.iterrows():
    print(f"✅ Successfully consumed record {row['record_id']}: Topic -> {row['topic']} | Status -> {row['status']}")

print("🎯 Stage 1 Complete: Ingestion, Processing, and Consumption simulations are fully operational!")
```

```python
from pydantic import BaseModel, ValidationError

print("🔒 Stage 1.5: Data Contract Validation using Pydantic")

# Define Data Contract Schema
class DataContract(BaseModel):
    record_id: int
    topic: str
    status: str
    value: float

# Validate data locally against the contract
validated_records = []
for idx, row in df_stream.iterrows():
    try:
        # Validate each row dictionary against the Pydantic model
        valid_data = DataContract(**row.to_dict())
        validated_records.append(valid_data.model_dump())
        print(f"✅ Contract Validated Successfully for Record ID: {valid_data.record_id}")
    except ValidationError as e:
        print(f"❌ Validation Failed for row {idx}: {e}")

print("🎯 Data Contract check completed successfully without external broker errors!")
```

```python
# تثبيت المكتبات الأساسية للبحث الهجين
!pip install -q sentence-transformers rank-bm25

import numpy as np
from sentence_transformers import SentenceTransformer
from rank_bm25 import BM25Okapi

# 1. تجهيز النصوص (Documents Chunking)
documents = [
    "معمارية هندسة البيانات الحديثة تعتمد على طبقات البرونز والفضة والذهب.",
    "نظام كافكا يستخدم لمعالجة تدفقات البيانات الحية وضمان موثوقيتها.",
    "يمنع دخول البيانات التالفة إلى النظام Pydantic التحقق من المخطط عبر.",
    "تعتمد على قواعد البيانات المنتجة للبحث الدقيق RAG الذكاء الاصطناعي وأنظمة."
]

print(f"✅ تم تقطيع المستندات إلى {len(documents)} أجزاء (Chunks)")

# 2. توليد التضمينات (Embeddings) والبحث الدلالي
model = SentenceTransformer('all-MiniLM-L6-v2')
doc_embeddings = model.encode(documents)

# محاكاة قاعدة بيانات متجهة للبحث الدلالي
def vector_search(query, top_k=2):
    query_embedding = model.encode([query])[0]
    # حساب التشابه (Cosine Similarity)
    similarities = np.dot(doc_embeddings, query_embedding) / (
        np.linalg.norm(doc_embeddings, axis=1) * np.linalg.norm(query_embedding)
    )
    top_indices = np.argsort(similarities)[::-1][:top_k]
    return [(documents[i], similarities[i]) for i in top_indices]

# 3. محاكاة البحث الهجين (Hybrid Search: Vector + BM25)
tokenized_docs = [doc.split(" ") for doc in documents]
bm25 = BM25Okapi(tokenized_docs)

def hybrid_search(query):
    print(f"\n--- نتائج البحث الهجين للاستعلام: '{query}' ---")
    results = vector_search(query, top_k=2)
    for doc, score in results:
        print(f"النص: {doc} (الدرجة: {score:.4f})")

# تجربة البحث الهجين
hybrid_search("ما هي طبقات هندسة البيانات؟")
```
